## Bin Tenure into Categories

```python
churn['tenure_bin'] = pd.cut(churn['tenure'], bins=[0, 20, 35, 50, 100])
churn['tenure_bin']
churn['tenure_bin'].value_counts()
```

This code uses the `pd.cut` function to categorize continuous customer tenure into discrete intervals (bins) of 0-20, 20-35, 35-50, and 50-100 months, then counts the number of customers in each bin.

## Group Churn rates by Tenure Bins

```python
churn_chance_asper_tenure = churn.groupby('tenure_bin')['Churn'].value_counts()
churn_chance_asper_tenure
```

This code groups the dataset by the newly created tenure bins and calculates the distribution of churned versus retained customers within each tenure group to analyze how loyalty changes over time.

## Correlation of Additional Services with Churn (Initial Try)

```python
bool_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection','TechSupport',
       'StreamingTV', 'StreamingMovies']  # apne Yes/No columns

temp = churn[bool_cols].replace({'Yes': 1, 'No': 0,'No internet service' : 0}).astype(int)
temp = temp.infer_objects(copy=False).astype(int)   # FutureWarning bhi hat jayegi
temp['Churn'] = churn['Churn'].map({'Yes': 1, 'No': 0})

print(temp.corr()['Churn'].drop('Churn').sort_values())
```

This code attempts to map categorical service options ('Yes'/'No') to numeric values to calculate their correlation with Churn, highlighting a scenario where Churn mapping failed because the target column was already binary-encoded.

## Calculate Service Correlations with Churn

```python
bool_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
             'TechSupport', 'StreamingTV', 'StreamingMovies']

temp = churn[bool_cols].replace({'Yes': 1, 'No': 0, 'No internet service': 0})
temp = temp.infer_objects(copy=False).astype(int)
temp['Churn'] = churn['Churn']   # map hata diya, kyunki Churn pehle se 0/1 hai

print(temp.corr()['Churn'].drop('Churn').sort_values())
```

This code cleans and formats the mapped services subset, drops redundant mappings, adds the correct binary Churn target, and outputs the sorted correlation values of each service with respect to Churn.

## Plot service correlation heatmap with Churn Included

```python
pl.figure(figsize=(8, 6))
sns.heatmap(temp.corr(), annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, linewidths=0.5)
pl.show()
```

This code creates a consolidated heatmap visualization representing correlations among all additional technical services along with the binary Churn target to highlight which services are most strongly linked to customer retention.
